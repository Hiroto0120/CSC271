### Web Scraping Considerations

Let's start by trying to read the HTML from the Wikipedia page mentioned:

````python
import pandas as pd
import requests

url = 'https://en.wikipedia.org/wiki/List_of_countries_and_dependencies_by_population'

response = requests.get(url)
response.encoding = response.apparent_encoding

if response.status_code == 200:
    print(response.text)
else:
    print(f'The request was not successful: status code {response.status_code}')
````

Our request was unsuccessful! When we look up the status code (https://requests.readthedocs.io/en/latest/api/#status-code-lookup), we see that code 403 means forbidden.  Our request was received, but denied! The Wikipedia server detected that the request was coming from a program (not a human via a web browser) and refused it.

Wikipedia wants to deter the scraping of its pages. Automated scraping, especially frequent requests, can put heavy load on servers, slow down the site, and increase costs. Wikipedia also provides another way (called an API) to access data, which we'll explore next week and which is preferred over scraping.

Before we tried to scrape, we should have checked whether it is permitted.

### Determining whether to scrape

#### **1. Check whether an API exists.**

Some sites provide ways to extract data from them called APIs. If an API is provided, that is the preferred mechanism for extracting the data from the site and it should be used rather than scraping. We'll discuss APIs further next week.

#### **2. Respect privacy and copyright**

We also need to consider the content of the site. 

**Avoid scraping person or sensitive data.** Names, emails, phone numbers and other private details about people are sensitive. Even if the data is publicly available, collecting it systematicall can violate laws.

**Web content is often copyrighted.** Articles, images, and other content may be protected by copyright even if publicly available. Do not republish scraped content in ways that violate copyright or licensing rules. Check the site's license. For example, Wikipeida uses Creative Commons licenses that aloow for reuse with attribution if the licensing rules are followed.

#### **3. Check the site's `robots.txt` file.**

Most sites will have a `robots.txt` file at the root of their URL (e.g., https://en.wikipedia.org/robots.txt). 

This file provides information about what is permitted and forbidden. For example, Wikipedia indicates that all bots (programs) related to Google ads should not access their pages:

```
# advertising-related bots:
User-agent: Mediapartners-Google*
Disallow: /
```
These rules are conventions that are not enforced technically, but well-behaved bots are expected to follow them.

Here is another rule that applies to all programs (the * means all):

```
User-agent: *
Disallow: /w/
Allow: /wiki/
```

It says that programs should not access `/w/` pages (MediaWiki pages and admin URLs), but that they can access `/wiki/` pages (normal article pages).

Even if scraping is allowed, we should avoid generating a large number of request in a short time frame and avoid scraping during peak times.

#### **4. Check the Terms of Use**

Most websites have a Terms of Use policy that we should read and follow . Here is Wikipedia's detailed Terms of Use page:
https://foundation.wikimedia.org/wiki/Policy:Terms_of_Use

Within that page, it links to a User-Agent Policy:
https://foundation.wikimedia.org/wiki/Policy:Wikimedia_Foundation_User-Agent_Policy

That policy says **'As of February 15, 2010, Wikimedia sites require a HTTP User-Agent header for all requests.'** In other words, any programs that request data must include identifying information. It even includes examples of how to do that in different programming languages including Python.

We will update our request to include that information and the server will successfully respond to our request this time:
