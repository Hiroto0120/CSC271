## Task 1: Query the Database

#### A) Write a query that returns the names of the coldest months and the number of cities for which each month is the coldest.

```python
query = '''
SELECT coldest_month, COUNT(*) AS city_count
FROM temperature
GROUP BY coldest_month;
'''  

coldest_counts = w11_helpers.run_sql(conn, query)
coldest_counts
```

#### B) Write a query that returns the average snowfall for each province. Include only cities that have corresponding precipitation data.

```python
query = '''
SELECT c.province, AVG(p.snow) AS avg_snowfall
FROM cities c
JOIN precipitation p USING (city)
GROUP BY c.province;
'''  

avg_snow_province = w11_helpers.run_sql(conn, query)
avg_snow_province
```

#### C) Write a query that returns the provinces that have more than one city in the `cities` table and the number of cities that each province has in that table.

```python

query = '''
SELECT province, COUNT(*) AS num_cities
FROM cities
GROUP BY province
HAVING COUNT(*) > 1;
'''  

multi_city_provinces = w11_helpers.run_sql(conn, query)
multi_city_provinces
```

#### D) Write a query that returns provinces and the average snowfall for cities in those provinces that have at least 50 days with precipitation. Include only provinces where the average snowfall is greater than 200.

```python
query = '''
SELECT c.province, AVG(p.snow) AS avg_snowfall
FROM cities c
JOIN precipitation p USING (city)
WHERE p.days_with_precip >= 50
GROUP BY c.province
HAVING AVG(p.snow) > 200;
'''

prov_avg_snow_snowiest = w11_helpers.run_sql(conn, query)
prov_avg_snow_snowiest
```

#### E) Write a query that returns all cities from the cities table and their snowfalls. If a city from the cities table has no precipitation data, include it with a missing snowfall value.

```python
query = ''' 
SELECT c.city, p.snow
FROM cities c
LEFT JOIN precipitation p USING (city);
'''

city_snow = w11_helpers.run_sql(conn, query)
city_snow
```
