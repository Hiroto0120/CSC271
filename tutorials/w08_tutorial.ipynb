Additionally, the database has these foreign keys:

- In the `precipitation` table, the `city` column must reference the `city` column in the `cities` table.

After completing this week’s lecture prep, you’ll see that the tables can also be described using schema notation:

<pre>temperature(
  <u>city</u>: TEXT, average_high: FLOAT, average_low: FLOAT,
  coldest_month: TEXT, avg_high_coldest_month: FLOAT, avg_low_coldest_month: FLOAT,
  warmest_month: TEXT, avg_high_warmest_month: FLOAT, avg_low_warmest_month: FLOAT)</pre>

Complete the three code blocks below to create the three tables with the constraints specified above:

```python
# TODO: write the create table SQL statement for the cities table
sql_cities = '''
CREATE TABLE cities (
    city TEXT PRIMARY KEY,
    province TEXT
)
'''

# Call on the helper function to add the table via the open connection
w08_helpers.add_table(conn, sql_cities)
```

```python
# TODO: add the precipitation table to the database
sql_precipitation = '''
CREATE TABLE precipitation (
    city TEXT PRIMARY KEY,
    snow FLOAT,
    total_precip FLOAT,
    days_with_precip INT,
    FOREIGN KEY (city) REFERENCES cities(city)
)
'''

w08_helpers.add_table(conn, sql_precipitation)
```

```python
# TODO: add the temperature table to the database
sql_temperature = '''
CREATE TABLE temperature (
    city TEXT PRIMARY KEY,
    average_high FLOAT,
    average_low FLOAT,
    coldest_month TEXT,
    avg_high_coldest_month FLOAT,
    avg_low_coldest_month FLOAT,
    warmest_month TEXT,
    avg_high_warmest_month FLOAT,
    avg_low_warmest_month FLOAT,
    FOREIGN KEY (city) REFERENCES cities(city)
)
'''

w08_helpers.add_table(conn, sql_temperature)
```
